```python
!pip install faiss-cpu
import torch
import faiss
import numpy as np
from datasets import load_dataset
from transformers import AutoTokenizer, AutoModelForCausalLM
from sentence_transformers import SentenceTransformer

# 1. Load the dataset (using the 'label' field for plant disease info)
dataset = load_dataset("A2H0H0R1/plant-disease-new")
texts = [str(item["label"]) for item in dataset["train"]]

# 2. Create text embeddings using a SentenceTransformer model
text_embedder = SentenceTransformer("all-MiniLM-L6-v2")
text_embeddings = text_embedder.encode(texts, convert_to_numpy=True)

# 3. Build a FAISS index for fast similarity search
embedding_dim = text_embeddings.shape[1]
index = faiss.IndexFlatL2(embedding_dim)
index.add(text_embeddings)

def retrieve_similar_text(query, k=5):
    query_embedding = text_embedder.encode(query, convert_to_numpy=True).reshape(1, -1)
    distances, indices = index.search(query_embedding, k)
    return [texts[i] for i in indices[0]]

# 4. Initialize Llama-2 for text generation (chatbot responses)
model_id = "meta-llama/Llama-2-7b-chat-hf"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id, device_map="auto")

def generate_answer(prompt, max_length=150):
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    with torch.no_grad():
        outputs = model.generate(**inputs, max_new_tokens=max_length)
    return tokenizer.decode(outputs[0], skip_special_tokens=True)

def chatbot_response(user_query):
    # Retrieve relevant plant disease information from our database
    retrieved_info = retrieve_similar_text(user_query)
    # Compose a prompt using retrieved information and the user query
    prompt = (
        f"Based on the following plant disease information: {', '.join(retrieved_info)}. "
        f"Answer the question: {user_query}"
    )
    return generate_answer(prompt)

# Corrected conditional statement: Using __name__ instead of _name_
if __name__ == "__main__":
    print("Welcome to the Plant Disease Chatbot. Type 'exit' to quit.")
    while True:
        user_input = input("User: ")
        if user_input.lower() in ["exit", "quit"]:
            break
        response = chatbot_response(user_input)
        print("Bot:", response)
```

```python
!git clone https://github.com/facebookresearch/segment-anything.git
!cd segment-anything
!pip install -e .
!huggingface-cli login
!pip install faiss-cpu
!pip install datasets
!pip install git+https://github.com/facebookresearch/segment-anything.git
!pip install opencv-python
!pip install sentence-transformers
!pip install accelerate
!pip install transformers --upgrade
import torch
import faiss
import numpy as np
import cv2
import matplotlib.pyplot as plt
from accelerate import init_empty_weights, load_checkpoint_and_dispatch
from datasets import load_dataset
from transformers import AutoTokenizer, AutoModelForCausalLM, AutoConfig
from sentence_transformers import SentenceTransformer
from segment_anything import sam_model_registry, SamPredictor
from PIL import Image
import requests
import os

# ---------- 1. Load dataset and FAISS index ----------
dataset = load_dataset("A2H0H0R1/plant-disease-new")
texts = [str(item["label"]) for item in dataset["train"]]

text_embedder = SentenceTransformer("all-MiniLM-L6-v2")
text_embeddings = text_embedder.encode(texts, convert_to_numpy=True)

embedding_dim = text_embeddings.shape[1]
index = faiss.IndexFlatL2(embedding_dim)
index.add(text_embeddings)

def retrieve_similar_text(query, k=5):
    query_embedding = text_embedder.encode(query, convert_to_numpy=True).reshape(1, -1)
    distances, indices = index.search(query_embedding, k)
    return [texts[i] for i in indices[0]]

# ---------- 2. Load LLaMA-2 model ----------
model_id = "meta-llama/Llama-2-7b-chat-hf"
tokenizer = AutoTokenizer.from_pretrained(model_id)

# Load the model directly with the desired configuration
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    torch_dtype=torch.float16,  # Use float16 for memory efficiency
    device_map="auto",  # Automatically map to available devices
    # load_in_8bit=True # Enable 8-bit quantization if supported
)

# Download the model weights to a local checkpoint file if it doesn't exist
checkpoint_file = "llama-2-7b-chat-hf.bin"  # You can change the filename
if not os.path.exists(checkpoint_file):
    print("Downloading model weights...")
    # Save the model using safe_serialization=True to avoid the RuntimeError
    model.save_pretrained(checkpoint_file, safe_serialization=True)


def generate_answer(prompt, max_length=150):
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    with torch.no_grad():
        outputs = model.generate(**inputs, max_new_tokens=max_length)
    return tokenizer.decode(outputs[0], skip_special_tokens=True)

# ---------- 3. Load Segment Anything Model (SAM) ----------
sam_checkpoint = "sam_vit_h_4b8939.pth"  # Download from: https://github.com/facebookresearch/segment-anything#model-checkpoints
if not os.path.exists(sam_checkpoint):
  sam_checkpoint_url = "https://dl.fbaipublicfiles.com/segment_anything/sam_vit_h_4b8939.pth"
  response = requests.get(sam_checkpoint_url, stream=True)
  with open(sam_checkpoint, 'wb') as f:
    for chunk in response.iter_content(chunk_size=1024):
      if chunk:
        f.write(chunk)

sam = sam_model_registry["vit_h"](checkpoint=sam_checkpoint)
sam.to("cuda" if torch.cuda.is_available() else "cpu")
predictor = SamPredictor(sam)

def segment_leaf_image(image_path):
    image = cv2.imread(image_path)
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    predictor.set_image(image_rgb)

    # Auto mask generation (grab a central point for simplicity)
    h, w, _ = image.shape
    input_point = np.array([[w // 2, h // 2]])
    input_label = np.array([1])

    masks, _, _ = predictor.predict(
        point_coords=input_point,
        point_labels=input_label,
        multimask_output=False,
    )

    # Draw mask on image
    mask = masks[0]
    annotated = image_rgb.copy()
    annotated[mask] = [255, 0, 0]  # Highlight region in red

    return Image.fromarray(annotated), "Possible disease region identified from leaf"

# ---------- 4. Multimodal chatbot logic ----------
def multimodal_response(user_text=None, image_path=None):
    retrieved_info = []

    # If text is provided, do semantic retrieval
    if user_text:
        retrieved_info.extend(retrieve_similar_text(user_text))

    # If image is provided, perform segmentation
    if image_path:
        annotated_img, image_info = segment_leaf_image(image_path)
        annotated_img.show()
        retrieved_info.append(image_info)

    # Compose prompt
    prompt = (
        f"Based on the following plant disease information: {', '.join(retrieved_info)}. "
        f"{'Answer the question: ' + user_text if user_text else 'Provide insights based on the image.'}"
    )
    return generate_answer(prompt)

# ---------- 5. CLI entry point ----------
if __name__ == "__main__":
    print("🌿 Welcome to the Multimodal Plant Disease Chatbot 🌿")
    print("Type your question or provide an image path. Type 'exit' to quit.\n")

    while True:
        mode = input("Choose mode (text/image/both): ").lower()
        if mode == "exit":
            break

        text_input, image_input = None, None
        if mode in ["text", "both"]:
            text_input = input("Enter your plant-related question: ")
        if mode in ["image", "both"]:
            image_input = input("Enter image file path: ")

        try:
            response = multimodal_response(user_text=text_input, image_path=image_input)
            print("🧠 Bot:", response)
        except Exception as e:
            print("❌ Error:", e)
```
